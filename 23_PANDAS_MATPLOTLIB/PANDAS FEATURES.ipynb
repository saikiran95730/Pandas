## 7. Merging and joining

```python
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Ravi', 'Ram', 'Priya']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Salary': [30000, 25000, 40000]
})
```

```python
pd.merge(df1, df2)
```

```python
pd.merge(df1, df2, on='ID')
```

```python
df1.join(df2.set_index('ID'), on='ID')
```

```python
merge()
    df1.ID ↔ df2.ID 
    Column matched with column
   merging inside

join()
    df1.ID ↔ df2.index
    Column matched with index
   joining df1 with df2 means:
     in df1 -- ID numbers-1,2,3 -- total - 3
     in df2 -- total indexes 3
     both are matching so they join together
```

```python
pd.concat([df1, df1])
```

## 8. Reshaping and pivoting

```python
sales = pd.DataFrame({
    'Name': ['Ravi', 'Ravi', 'Ram', 'Ram'],
    'Year': [2025, 2026, 2025, 2026],
    'Sales': [100, 200, 150, 250]
})
```

```python
sales.pivot(index='Name', columns='Year', values='Sales')
```

```python
** 
.pivot() - converts unique values from one column into separate separate columns.
```

```python
sales.pivot_table(
    index='Name',
    values='Sales',
    aggfunc='sum'
)
```

```python
** 
.pivot_table()
    Groups sales by Name indexes.
    Adds the sales values using sum.
```

```python
sales.melt(id_vars='Name')
```

```python
**
.melt() — convert columns into rows
    -Name remains unchanged.
    -Year columns become rows under variable.
    -Their values go under value.
```

```python
sales.set_index(['Name', 'Year']).stack()
```

```python
** 
.stack() —move columns into an inner row index(Years)
 What happened?

 The year columns were moved into the row index, creating a more vertically arranged result.
```

```python
sales.set_index(['Name', 'Year']).unstack()
```

```python
**
.unstack() — move an inner row index into columns(Years)
```
